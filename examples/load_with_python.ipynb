# Load `dental-notes-benchmark` with Python

This notebook shows a simple way to load the JSONL dataset, inspect key fields, and compute a few descriptive statistics.

```python
from pathlib import Path
import json
import pandas as pd

# Update this path if needed
dataset_path = Path("../data/dentalnote_bench_20000_icd10_internalproc_v20.jsonl")
dataset_path
```

```python
records = []
with dataset_path.open("r", encoding="utf-8") as f:
    for i, line in enumerate(f):
        records.append(json.loads(line))
        if i >= 999:  # load first 1,000 rows for quick exploration
            break

len(records)
```

```python
df = pd.DataFrame(records)
df.head(3)
```

```python
df[["note_id", "split", "visit_type", "specialty", "summary_type"]].head(10)
```

```python
df["visit_type"].value_counts()
```

```python
df["specialty"].value_counts()
```

```python
df["note_length"] = df["input_note"].astype(str).str.len()
df["summary_length"] = df["summary_SOAP"].astype(str).str.len()
df[["note_length", "summary_length"]].describe()
```

```python
# Example: flatten diagnoses for quick inspection
diag_rows = []
for row in records:
    for d in row.get("diagnoses", []):
        diag_rows.append({
            "note_id": row.get("note_id"),
            "visit_type": row.get("visit_type"),
            "specialty": row.get("specialty"),
            "diagnosis_code": d.get("diagnosis_code"),
            "diagnosis_text": d.get("diagnosis_text"),
        })

diag_df = pd.DataFrame(diag_rows)
diag_df.head()
```

```python
diag_df["diagnosis_code"].value_counts().head(20)
```

## Notes

- The example only loads the first 1,000 rows for speed.
- For full-scale training or analysis, stream the JSONL file instead of loading everything into memory.
- The dataset is synthetic and intended for benchmarking and research use.
